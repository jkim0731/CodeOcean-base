```python
def read_all_saved_files(save_dir):
    loaded_vars = {}
    for fn in save_dir.glob("*.npy"):
        print(f"Loading {fn}...")
        loaded_vars[fn.stem] = np.load(fn, allow_pickle=True)
    for fn in save_dir.glob("*.csv"):
        print(f"Loading {fn}...")
        loaded_vars[fn.stem] = pd.read_csv(fn)
    return loaded_vars
```
