```python
import jax
import jax.numpy as jnp
import numpy as np
from statsmodels.robust.norms import RobustNorm, TukeyBiweight
```

```python
from dev import AsymmetricTukeyBiweight, AsymmetricTukeyBiweight_jax, TukeyBiweight_jax
```

```python
x = np.arange(-10,10,.2)
for a in (AsymmetricTukeyBiweight(), AsymmetricTukeyBiweight_jax(), TukeyBiweight_jax()):
    print(
        (np.allclose(a.rho(x), TukeyBiweight().rho(x)),
        np.allclose(a.psi(x), TukeyBiweight().psi(x)),
        np.allclose(a.weights(x), TukeyBiweight().weights(x)),
        np.allclose(a.psi_deriv(x), TukeyBiweight().psi_deriv(x)))
    )
```

```python
for a in (AsymmetricTukeyBiweight(1,1), AsymmetricTukeyBiweight_jax(1,1), TukeyBiweight_jax(1)):
    print(
        (np.allclose(a.rho(x), TukeyBiweight(1).rho(x)),
        np.allclose(a.psi(x), TukeyBiweight(1).psi(x)),
        np.allclose(a.weights(x), TukeyBiweight(1).weights(x)),
        np.allclose(a.psi_deriv(x), TukeyBiweight(1).psi_deriv(x)))
    )
```

```python
for a in (AsymmetricTukeyBiweight_jax(2,1), ):
    print(
        (np.allclose(a.rho(x), AsymmetricTukeyBiweight(2,1).rho(x)),
        np.allclose(a.psi(x), AsymmetricTukeyBiweight(2,1).psi(x)),
        np.allclose(a.weights(x), AsymmetricTukeyBiweight(2,1).weights(x)),
        np.allclose(a.psi_deriv(x), AsymmetricTukeyBiweight(2,1).psi_deriv(x)))
    )
```
